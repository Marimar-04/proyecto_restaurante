```python
from my_library.data_processing import DataProcessor
from my_library.analysis import Analysis
from my_library.visualization import Visualization
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Procesamiento
json_dir = os.path.abspath('./data')
processor = DataProcessor(json_dir)

data = processor.data

analysis = Analysis()
promedios = {}
for items in data['data']:
    print(items)
    promedios[items['name']] = analysis.calculate_average_prices(items['menu']['card'])

df = pd.DataFrame(promedios)

display(df)
# Gráfico de calor
plt.figure(figsize=(12, 8))
sns.heatmap(df, annot=True, fmt=".2f", cmap="coolwarm", cbar_kws={'label': 'Precio Promedio'})
plt.title("Precios Promedio por Categoría y Restaurante")
plt.ylabel("Restaurante")
plt.xlabel("Categoría")
plt.show()

```
